### Contexto

1. Identificar las causas principales del abandono de usuarios (*Churn Rate*).
2. Determinar la relación entre el tipo de dispositivo/contenido y el tiempo total de reproducción (*Engagement Index*).

### 2) Carga de los datos

```python
%pip install dash
## Todas las librerias necesarias
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
from dash import Dash, dcc, html, Input, Output
import plotly.io as pio
pio.renderers.default = "vscode"
from IPython.display import HTML

print("Librerías cargadas correctamente.")
```

# Datasets separados

```python
## Cargar los datos
df_tv = pd.read_csv('../data/netflix_tv_shows_detailed_up_to_2025.csv')
df_movies = pd.read_csv('../data/netflix_movies_detailed_up_to_2025.csv')

print('============================= TV Shows =============================')
display(df_tv)
print('=============================== Movies ===============================')
display(df_movies)
```

# DataSets fusionados

```python
## Data wrangling: homologar e integrar
df_tv['tipo_contenido'] = 'TV Show'
df_movies['tipo_contenido'] = 'Movie'

## Fusionamos los datasets aun manteniendo variables comunes criticas para el negocio
cols_comunes = ['show_id', 'tipo_contenido', 'title', 'country', 'date_added',
                'release_year', 'genres', 'popularity', 'vote_average']
df_master_raw = pd.concat([df_tv[cols_comunes], df_movies[cols_comunes]], ignore_index=True)

## Hacemos una copia profunda para no alterar la data original
df_master_processed = df_master_raw.copy()

print("=================== Dataset antes del Preprocesamiento ===================\n")
print(f"Dimensión del dataset crudo: {df_master_processed.shape[0]} filas x {df_master_processed.shape[1]} columnas")
print("\nValores nulos existentes por columna:")
print(df_master_processed.isnull().sum())

display(df_master_raw)

## imputacion de valores nulos
cols_texto = ['country', 'genres']
df_master_processed[cols_texto] = df_master_processed[cols_texto].fillna('Unknown')

## Transformacion de tipos de datos (fechas)
df_master_processed['date_added'] = pd.to_datetime(df_master_processed['date_added'], errors='coerce')

## Eliminar registros donde la fecha no se pudo procesar (si los hay)
df_master_processed = df_master_processed.dropna(subset=['date_added'])

## Feature Engineering: Nuevas variables creadas para el analisis
df_master_processed['año_agregado'] = df_master_processed['date_added'].dt.year
## Extraemos solo el primer genero antes de la coma
df_master_processed['genero_principal'] = df_master_processed['genres'].apply(lambda x: str(x).split(',')[0].strip())

## eliminar posibles duplicados
df_master_processed = df_master_processed.drop_duplicates(subset=['show_id'])

print("\n================== Dataset despúes del Preprocesamiento ==================\n")
print("Preprocesamiento finalizado con éxito.")
print(f"Dimensión del dataset limpio: {df_master_processed.shape[0]} filas x {df_master_processed.shape[1]} columnas")
print("\nValores nulos restantes por columna:")
print(df_master_processed.isnull().sum())

df_master_processed
```

# **II** - Top 10 Generos por popularidad promedio


*   Para el negocio, es vital saber que generos generan mas interes ("Popularidad"). Este grafico de barras horizontales responde a eso.




```python
df_generos = df_master_processed.groupby(['genero_principal', 'tipo_contenido'])['popularity'].mean().reset_index()

## Filtramos los 10 generos principales en general para no saturar el grafico
top_10_generos = df_master_processed['genero_principal'].value_counts().head(10).index
df_generos_top = df_generos[df_generos['genero_principal'].isin(top_10_generos)]

fig_bar = px.bar(
    df_generos_top,
    x='popularity',
    y='genero_principal',
    color='tipo_contenido',
    barmode='group',
    title='Nivel de Interacción: "Popularidad Promedio" en el Top 10 de Géneros',
    labels={'popularity': 'Popularidad Promedio', 'genero_principal': 'Género Principal'},
    color_discrete_map={'Movie': '#EE9B00', 'TV Show': '#005F73'}
)

## Ordenamos de mayor a menor para facilitar la comparacion, efectuando principio de jerarquia visual
fig_bar.update_layout(yaxis={'categoryorder':'total ascending'}, template='plotly_white')
HTML(fig_bar.to_html(include_plotlyjs='cdn'))
```

## Justificacion del Analisis Exploratorio (EDA) y Decisiones de Diseño

Para abordar los objetivos estrategicos de StreamView Analytics, el analisis exploratorio se diseñó bajo principios de percepcion visual orientados a minimizar la carga cognitiva de la audiencia (C-Level y Product Managers) y facilitar la rapida identificacion de patrones de consumo.

### Selección de Representaciones Gráficas
* **Gráfico de Líneas (Evolución Temporal):** Se utilizo para representar el volumen de contenido agregado por año. La conexión continua entre los puntos facilita la percepción de tendencias temporales, permitiendo identificar rápidamente los periodos de mayor expansión del catálogo.
* **Gráfico de Barras Horizontales (Top Géneros por Popularidad):** Se seleccionó este formato para comparar datos categóricos con etiquetas largas (nombres de géneros). La orientación horizontal permite una lectura natural de izquierda a derecha, evitando inclinar el texto y mejorando la legibilidad en comparación con un gráfico de barras verticales o un gráfico circular.
* **Gráfico de Dispersión (Calificación vs. Popularidad):** Es la representación óptima para revelar correlaciones entre dos variables cuantitativas. Se aplicó un nivel de opacidad (canal alfa) a los puntos para mitigar el problema de superposición (*overplotting*) y visualizar correctamente las áreas de mayor densidad de contenido.

### Atributos Visuales y Principios de Percepción
* **Uso Estratégico del Color (Contraste y Categorización):** Se aplicó una paleta de colores intencionada para diferenciar claramente los formatos de contenido, asignando el color rojo a las Películas (Movies) y el color verde a las Series (TV Shows). Esta decisión aprovecha el procesamiento pre-atentivo del cerebro, permitiendo a la audiencia distinguir, agrupar y comparar el rendimiento de ambas categorías de manera inmediata y sin esfuerzo cognitivo adicional.
* **Simplicidad y Jerarquía:** Se eliminaron las líneas de cuadrícula innecesarias y se ordenaron las barras de mayor a menor popularidad. Esto responde al principio de jerarquía visual, estructurando la información para que el cerebro procese primero los datos de mayor impacto para el negocio.

### Principales Hallazgos
* Existe una clara preferencia de la audiencia por un grupo reducido de géneros, lo que indica dónde debería concentrarse el presupuesto de adquisición para mejorar la retención.
* La dispersión muestra que los títulos con calificaciones más altas no siempre son los que generan mayor volumen de interacción (popularidad), sugiriendo que el *engagement* depende más del formato y género que de la aclamación crítica.

## 4) Dashboard Interactivo

*   Este bloque toma el dataframe df_master_raw (que ya está limpio en la memoria de Colab), crea un panel con KPIs dinámicos, agrega filtros por tipo de contenido y año, y vincula todo para que se actualice automáticamente.



```python
## Inicializamos la aplicacion Dash
app = Dash(__name__)

## Definimos el diseño del Dashboard. (el layout)
app.layout = html.Div(style={'font-family': 'Arial, sans-serif', 'padding': '20px', 'backgroundColor': '#f9f9f9'}, children=[

    ## encabezado
    html.H1("🎬 StreamView Analytics - Panel de Control Interactivo", style={'textAlign': 'center', 'color': '#9B82C3'}),
    html.P("Explora el comportamiento del contenido, popularidad y calificaciones.", style={'textAlign': 'center'}),
    html.Hr(),

    ## panel de filtros
    html.Div([
        ## filtro de tipo (izquierda)
        html.Div([
            html.Label("Filtro por Tipo de Contenido:", style={'fontWeight': 'bold'}),
            dcc.Dropdown(
                id='filtro-tipo',
                options=[
                    {'label': 'Todos', 'value': 'Todos'},
                    {'label': 'Películas (Movies)', 'value': 'Movie'},
                    {'label': 'Series (TV Shows)', 'value': 'TV Show'}
                ],
                value='Todos',
                clearable=False,
            )
        ], style={'width': '45%', 'display': 'inline-block', 'verticalAlign': 'top'}),

        ## filtro de Años (derecha)
        html.Div([
            html.Label("Rango de Años (Agregado):", style={'fontWeight': 'bold'}),
            dcc.RangeSlider(
                id='filtro-anos',
                min=2010, ##esto se ajusta esto segun el año minimo del dataset
                max=2024, ## esto se ajusta esto segun el año maximo del dataset
                step=1,
                marks={i: str(i) for i in range(2010, 2025, 2)}, ## esto por defecto muestra los numeros cada 2 años
                value=[2015, 2023] ## Valores seleccionados por defecto
            )
        ], style={'width': '45%', 'display': 'inline-block', 'marginLeft': '5%', 'verticalAlign': 'top'})
    ], style={'marginBottom': '30px'}),

    ## tarjetas de KPIs. indicadores clave
    html.Div(id='kpi-cards', style={'display': 'flex', 'justifyContent': 'space-around', 'marginBottom': '30px'}),

    ## graficos
    html.Div([
        dcc.Graph(id='grafico-tendencia', style={'width': '48%', 'display': 'inline-block'}),
        dcc.Graph(id='grafico-generos', style={'width': '48%', 'display': 'inline-block'})
    ])
])

## definimos la interactividad. callbacks
@app.callback(
    [Output('kpi-cards', 'children'),
     Output('grafico-tendencia', 'figure'),
     Output('grafico-generos', 'figure')],
    [Input('filtro-tipo', 'value'),
     Input('filtro-anos', 'value')]
)
def actualizar_dashboard(tipo_seleccionado, rango_anos):
    ## paso 1: filtrar los datos sobre el dataset !PROCESADO!
    ##Filtro tipo
    if tipo_seleccionado == 'Todos':
        df_filtrado = df_master_processed.copy()
    else:
        df_filtrado = df_master_processed[df_master_processed['tipo_contenido'] == tipo_seleccionado]

    ##Filtro por rango de años
    ano_min, ano_max = rango_anos
    df_filtrado = df_filtrado[(df_filtrado['año_agregado'] >= ano_min) & (df_filtrado['año_agregado'] <= ano_max)]

    ## paso2: calcular KPIs
    total_titulos = len(df_filtrado)
    pop_promedio = df_filtrado['popularity'].mean()
    rating_promedio = df_filtrado['vote_average'].mean()

    kpi_layout = [
        html.Div([
            html.H3("Total Títulos"),
            html.H2(f"{total_titulos:,}")
        ], style={'padding': '20px', 'backgroundColor': 'white', 'boxShadow': '2px 2px 5px #ccc', 'borderRadius': '10px', 'textAlign': 'center', 'width': '30%'}),
        html.Div([
            html.H3("Popularidad Promedio"),
            html.H2(f"{pop_promedio:.1f}")
        ], style={'padding': '20px', 'backgroundColor': 'white', 'boxShadow': '2px 2px 5px #ccc', 'borderRadius': '10px', 'textAlign': 'center', 'width': '30%'}),
        html.Div([
            html.H3("Calificación Promedio"),
            html.H2(f"{rating_promedio:.1f} / 10")
        ], style={'padding': '20px', 'backgroundColor': 'white', 'boxShadow': '2px 2px 5px #ccc', 'borderRadius': '10px', 'textAlign': 'center', 'width': '30%'})
    ]

    ## aplicamos el mapa de colores: movies/rojo, tv shows/verde
    color_map = {'Movie': '#E50914', 'TV Show': '#008000'}

    ## Paso 3:actualizar grafico de tendencia
    df_trend = df_filtrado.groupby(['año_agregado', 'tipo_contenido']).size().reset_index(name='cantidad')
    df_trend = df_trend[df_trend['año_agregado'] >= 2010]

    fig_trend = px.line(
        df_trend, x='año_agregado', y='cantidad', color='tipo_contenido',
        title='📈 Evolución del Contenido por Año',
        labels={'año_agregado': 'Año', 'cantidad': 'Volumen de Títulos', 'tipo_contenido': 'Formato'},
        color_discrete_map=color_map
    )
    fig_trend.update_layout(template='plotly_white')

    ## Actualizar grafico de generos
    ## calculamos el top 10 general y luego graficamos respetando el tipo y los colores
    df_generos = df_filtrado.groupby(['genero_principal', 'tipo_contenido'])['popularity'].mean().reset_index()
    top_10 = df_filtrado['genero_principal'].value_counts().head(10).index
    df_generos = df_generos[df_generos['genero_principal'].isin(top_10)]

    fig_bar = px.bar(
        df_generos, x='popularity', y='genero_principal', color='tipo_contenido',
        orientation='h', barmode='group',
        title='📊 Top 10 Géneros por Popularidad',
        labels={'popularity': 'Popularidad Promedio', 'genero_principal': 'Género', 'tipo_contenido': 'Formato'},
        color_discrete_map=color_map
    )
    fig_bar.update_layout(yaxis={'categoryorder':'total ascending'}, template='plotly_white')

    return kpi_layout, fig_trend, fig_bar

## linea que ejecuta la aplicacion dentro de la celda de Colab
if __name__ == '__main__':
    app.run(jupyter_mode="inline", port=8050)
```

## 4) Análisis Multivariado de Mercados y Justificación Técnica de Representaciones Visuales (IE25, IE27, IE28, IE29)

### Justificación de Diseño Visual y Percepción (Cleveland & McGill / Leyes de la Gestalt)
- **Accesibilidad Cromática (IE29):** Se reemplazó la codificación semafórica tradicional (rojo/verde) por una **paleta corporativa segura para daltonismo (Deuteranopía/Protanopía)**: **Azul Petróleo (`#005F73`)** para *TV Shows* y **Ámbar Cálido (`#EE9B00`)** para *Movies*, garantizando contraste tanto en matiz como en luminancia sobre fondo blanco (`plotly_white`) para minimizar la carga cognitiva y maximizar el *data-ink ratio* (Tufte).
- **Jerarquía Perceptual (IE25, IE28):**
  1. **Gráfico de Líneas (Evolución Temporal):** Codifica series continuas (2010–2025) mediante posición en una escala común y pendiente, óptimo para detectar el estancamiento de volumen (~1.000 títulos/año por formato desde 2019) frente a las fluctuaciones de popularidad.
  2. **Gráfico de Barras Horizontales Agrupadas (Top 10 Géneros):** Aprovecha el atributo preatentivo de **longitud alineada**, facilitando la lectura de etiquetas categóricas y evidenciando que las series superan en **3,16 veces** la popularidad media de las películas (64,88 vs. 20,53 puntos).
  3. **Diagrama de Dispersión con Transparencia $\alpha = 0.4$ (Calificación vs. Popularidad):** Mitiga el *overplotting* en los 31.594 registros y demuestra empíricamente la baja correlación lineal global ($r = 0.052$), permitiendo segmentar el catálogo en **Cuadrantes Estratégicos** (Anclas de Retención, Contenido Viral/Reality y Joyas Ocultas).
  4. **Mapa de Calor Multivariado (País vs. Género):** Codifica mediante saturación cromática las intersecciones de mayor tracción internacional.

```python
## IV - Mapa de Calor: Popularidad Promedio por País Productor y Género Principal
df_master_processed['pais_principal'] = df_master_processed['country'].apply(lambda x: str(x).split(',')[0].strip())
top_paises = ['United States of America', 'Japan', 'South Korea', 'United Kingdom', 'China', 'Canada', 'France', 'India']
top_generos_hm = ['Drama', 'Comedy', 'Animation', 'Action', 'Reality', 'Crime', 'Documentary']

df_hm = df_master_processed[
    (df_master_processed['pais_principal'].isin(top_paises)) &
    (df_master_processed['genero_principal'].isin(top_generos_hm))
]
piv_hm = df_hm.pivot_table(index='pais_principal', columns='genero_principal', values='popularity', aggfunc='mean').fillna(0)

fig_hm = px.imshow(
    piv_hm,
    text_auto='.1f',
    aspect='auto',
    color_continuous_scale='YlGnBu',
    title='Mapa de Calor: Índice de Popularidad Promedio por País de Origen y Género Principal',
    labels=dict(x='Género Principal', y='País Productor', color='Popularidad Media')
)
fig_hm.update_layout(template='plotly_white')
HTML(fig_hm.to_html(include_plotlyjs='cdn'))
```

## 6) Diseño e Implementación del Dashboard Interactivo en Dash (Evaluación 2 + EP5)

A continuación se implementa la aplicación interactiva completa en **Dash y Plotly**, integrando:
- **4 Indicadores Clave de Desempeño (KPIs dinámicos):** Catálogo Activo Filtrado, Índice de Engagement Promedio, Calificación Media de Usuarios (sobre títulos votados) y Cuota de Series.
- **4 Mecanismos de Filtrado Multidimensional:** Selector de Formato (`TV Show` / `Movie`), País Productor Principal, Género Principal y `RangeSlider` temporal (2010–2025).
- **Navegación Estructurada en 3 Pestañas Analíticas:** Panorama Ejecutivo, Engagement y Matriz Calidad-Tracción, y Mercados Internacionales.

```python
"""
Dashboard Interactivo Ejecutivo - StreamView Analytics (EP5)
Autores: Benjamin Patiño & Benjamin Vargas
Asignatura: Visualización de Datos (ADY1104) - Duoc UC
"""
import os
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from dash import Dash, dcc, html, Input, Output

# Carga resiliente del dataset maestro o de las fuentes originales
if os.path.exists('../data/streamview_master_processed.csv'):
    df_master_processed = pd.read_csv('../data/streamview_master_processed.csv')
elif os.path.exists('streamview_master_processed.csv'):
    df_master_processed = pd.read_csv('streamview_master_processed.csv')
else:
    df_tv = pd.read_csv('netflix_tv_shows_detailed_up_to_2025.csv')
    df_movies = pd.read_csv('netflix_movies_detailed_up_to_2025.csv')
    df_tv['tipo_contenido'] = 'TV Show'
    df_movies['tipo_contenido'] = 'Movie'
    cols = ['show_id', 'tipo_contenido', 'title', 'country', 'date_added',
            'release_year', 'genres', 'language', 'popularity', 'vote_count', 'vote_average']
    df_master_processed = pd.concat([df_tv[cols], df_movies[cols]], ignore_index=True)
    df_master_processed[['country', 'genres', 'language']] = df_master_processed[['country', 'genres', 'language']].fillna('Unknown')
    df_master_processed['date_added'] = pd.to_datetime(df_master_processed['date_added'], errors='coerce')
    df_master_processed = df_master_processed.dropna(subset=['date_added'])
    df_master_processed['año_agregado'] = df_master_processed['date_added'].dt.year
    df_master_processed['genero_principal'] = df_master_processed['genres'].apply(lambda x: str(x).split(',')[0].strip())
    df_master_processed['pais_principal'] = df_master_processed['country'].apply(lambda x: str(x).split(',')[0].strip())
    df_master_processed = df_master_processed.drop_duplicates(subset=['show_id'])

# Paleta corporativa accesible (Daltonismo-safe: Azul Petróleo vs Ámbar Cálido)
COLOR_MAP = {'TV Show': '#005F73', 'Movie': '#EE9B00'}

app = Dash(__name__)
app.title = "StreamView Analytics | Executive Dashboard"

paises_top = ['Todos'] + sorted(df_master_processed['pais_principal'].value_counts().head(15).index.tolist())
generos_top = ['Todos'] + sorted(df_master_processed['genero_principal'].unique().tolist())

app.layout = html.Div(style={'fontFamily': 'Segoe UI, Roboto, sans-serif', 'backgroundColor': '#F8F9FA', 'padding': '20px'}, children=[
    # Encabezado Corporativo
    html.Div(style={'backgroundColor': '#1D3557', 'color': 'white', 'padding': '20px 28px', 'borderRadius': '10px', 'marginBottom': '20px', 'boxShadow': '0 4px 10px rgba(0,0,0,0.1)'}, children=[
        html.H1("StreamView Analytics — Dashboard de Retención y Engagement", style={'margin': '0', 'fontSize': '26px'}),
        html.P("Solución Integral de Visualización de Datos para Decisiones C-Level y Product Managers | Consultores: Benjamin Patiño & Benjamin Vargas", style={'margin': '6px 0 0 0', 'color': '#A8DADC', 'fontSize': '14px'})
    ]),

    # Panel de Filtros Interactivos (Evaluación 2 + EP5)
    html.Div(style={'backgroundColor': 'white', 'padding': '18px', 'borderRadius': '10px', 'marginBottom': '20px', 'display': 'grid', 'gridTemplateColumns': 'repeat(4, 1fr)', 'gap': '18px', 'boxShadow': '0 2px 6px rgba(0,0,0,0.06)'}, children=[
        html.Div([
            html.Label("Formato de Contenido:", style={'fontWeight': 'bold', 'fontSize': '13px', 'color': '#1D3557'}),
            dcc.Dropdown(
                id='filtro-formato',
                options=[{'label': 'Todos los Formatos', 'value': 'Todos'},
                         {'label': 'Series (TV Show)', 'value': 'TV Show'},
                         {'label': 'Películas (Movie)', 'value': 'Movie'}],
                value='Todos', clearable=False
            )
        ]),
        html.Div([
            html.Label("País Productor Principal (Top 15):", style={'fontWeight': 'bold', 'fontSize': '13px', 'color': '#1D3557'}),
            dcc.Dropdown(
                id='filtro-pais',
                options=[{'label': p, 'value': p} for p in paises_top],
                value='Todos', clearable=False
            )
        ]),
        html.Div([
            html.Label("Género Principal:", style={'fontWeight': 'bold', 'fontSize': '13px', 'color': '#1D3557'}),
            dcc.Dropdown(
                id='filtro-genero',
                options=[{'label': g, 'value': g} for g in generos_top],
                value='Todos', clearable=False
            )
        ]),
        html.Div([
            html.Label("Rango de Año de Incorporación:", style={'fontWeight': 'bold', 'fontSize': '13px', 'color': '#1D3557'}),
            dcc.RangeSlider(
                id='filtro-anio',
                min=2010, max=2025, step=1,
                value=[2010, 2025],
                marks={yr: str(yr) for yr in range(2010, 2026, 3)}
            )
        ])
    ]),

    # Tarjetas de Indicadores Clave (KPIs)
    html.Div(id='kpi-container', style={'display': 'grid', 'gridTemplateColumns': 'repeat(4, 1fr)', 'gap': '18px', 'marginBottom': '20px'}),

    # Navegación por Pestañas (Storytelling & Exploración)
    dcc.Tabs(id='tabs-analisis', value='tab-ejecutivo', children=[
        dcc.Tab(label='1. Panorama Ejecutivo y Evolución Temporal', value='tab-ejecutivo', style={'fontWeight': 'bold'}, selected_style={'fontWeight': 'bold', 'borderTop': '3px solid #005F73'}),
        dcc.Tab(label='2. Engagement por Género y Matriz Calidad vs. Tracción', value='tab-engagement', style={'fontWeight': 'bold'}, selected_style={'fontWeight': 'bold', 'borderTop': '3px solid #EE9B00'}),
        dcc.Tab(label='3. Distribución Geográfica y Mercados Clave', value='tab-mercados', style={'fontWeight': 'bold'}, selected_style={'fontWeight': 'bold', 'borderTop': '3px solid #1D3557'})
    ]),

    html.Div(id='contenido-tabs', style={'marginTop': '18px'})
])

@app.callback(
    [Output('kpi-container', 'children'),
     Output('contenido-tabs', 'children')],
    [Input('filtro-formato', 'value'),
     Input('filtro-pais', 'value'),
     Input('filtro-genero', 'value'),
     Input('filtro-anio', 'value'),
     Input('tabs-analisis', 'value')]
)
def actualizar_dashboard(formato, pais, genero, rango_anio, tab_activa):
    dff = df_master_processed[
        (df_master_processed['año_agregado'] >= rango_anio[0]) &
        (df_master_processed['año_agregado'] <= rango_anio[1])
    ].copy()

    if formato != 'Todos':
        dff = dff[dff['tipo_contenido'] == formato]
    if pais != 'Todos':
        dff = dff[dff['pais_principal'] == pais]
    if genero != 'Todos':
        dff = dff[dff['genero_principal'] == genero]

    total_titulos = len(dff)
    pop_promedio = dff['popularity'].mean() if total_titulos > 0 else 0
    dff_votados = dff[dff['vote_count'] > 0]
    calif_promedio = dff_votados['vote_average'].mean() if len(dff_votados) > 0 else 0
    pct_series = (len(dff[dff['tipo_contenido'] == 'TV Show']) / total_titulos * 100) if total_titulos > 0 else 0

    card_style = {'backgroundColor': 'white', 'padding': '16px 20px', 'borderRadius': '10px', 'boxShadow': '0 2px 6px rgba(0,0,0,0.06)', 'borderLeft': '5px solid #005F73'}
    kpis = [
        html.Div(style=card_style, children=[
            html.Div("CATÁLOGO ACTIVO FILTRADO", style={'fontSize': '12px', 'color': '#6C757D', 'fontWeight': 'bold'}),
            html.Div(f"{total_titulos:,} títulos", style={'fontSize': '26px', 'fontWeight': 'bold', 'color': '#1D3557', 'marginTop': '4px'}),
            html.Div(f"Periodo {rango_anio[0]} - {rango_anio[1]}", style={'fontSize': '12px', 'color': '#2A9D8F'})
        ]),
        html.Div(style={**card_style, 'borderLeft': '5px solid #EE9B00'}, children=[
            html.Div("ÍNDICE DE ENGAGEMENT (POPULARIDAD)", style={'fontSize': '12px', 'color': '#6C757D', 'fontWeight': 'bold'}),
            html.Div(f"{pop_promedio:.1f} pts", style={'fontSize': '26px', 'fontWeight': 'bold', 'color': '#1D3557', 'marginTop': '4px'}),
            html.Div("Tracción media por título", style={'fontSize': '12px', 'color': '#E76F51'})
        ]),
        html.Div(style={**card_style, 'borderLeft': '5px solid #2A9D8F'}, children=[
            html.Div("CALIFICACIÓN MEDIA USUARIOS (VOTADOS)", style={'fontSize': '12px', 'color': '#6C757D', 'fontWeight': 'bold'}),
            html.Div(f"{calif_promedio:.2f} / 10", style={'fontSize': '26px', 'fontWeight': 'bold', 'color': '#1D3557', 'marginTop': '4px'}),
            html.Div(f"Sobre {len(dff_votados):,} títulos calificados", style={'fontSize': '12px', 'color': '#2A9D8F'})
        ]),
        html.Div(style={**card_style, 'borderLeft': '5px solid #1D3557'}, children=[
            html.Div("CUOTA DE SERIES (RETENCIÓN >90 DÍAS)", style={'fontSize': '12px', 'color': '#6C757D', 'fontWeight': 'bold'}),
            html.Div(f"{pct_series:.1f}% Series", style={'fontSize': '26px', 'fontWeight': 'bold', 'color': '#1D3557', 'marginTop': '4px'}),
            html.Div(f"{100 - pct_series:.1f}% Películas", style={'fontSize': '12px', 'color': '#6C757D'})
        ])
    ]

    if tab_activa == 'tab-ejecutivo':
        df_t = dff.groupby(['año_agregado', 'tipo_contenido']).agg(
            cantidad=('show_id', 'count'),
            popularidad=('popularity', 'mean')
        ).reset_index()
        fig_evol = px.line(
            df_t, x='año_agregado', y='cantidad', color='tipo_contenido', markers=True,
            title='Evolución Anual de Incorporaciones al Catálogo',
            labels={'año_agregado': 'Año de Ingreso', 'cantidad': 'Cantidad de Títulos', 'tipo_contenido': 'Formato'},
            color_discrete_map=COLOR_MAP, template='plotly_white'
        )
        fig_pop_t = px.line(
            df_t, x='año_agregado', y='popularidad', color='tipo_contenido', markers=True,
            title='Evolución del Índice de Popularidad Promedio por Formato',
            labels={'año_agregado': 'Año de Ingreso', 'popularidad': 'Popularidad Promedio', 'tipo_contenido': 'Formato'},
            color_discrete_map=COLOR_MAP, template='plotly_white'
        )
        contenido = html.Div(style={'display': 'grid', 'gridTemplateColumns': '1fr 1fr', 'gap': '18px'}, children=[
            html.Div(style={'backgroundColor': 'white', 'padding': '15px', 'borderRadius': '10px'}, children=[dcc.Graph(figure=fig_evol)]),
            html.Div(style={'backgroundColor': 'white', 'padding': '15px', 'borderRadius': '10px'}, children=[dcc.Graph(figure=fig_pop_t)])
        ])
    elif tab_activa == 'tab-engagement':
        top_g = dff['genero_principal'].value_counts().head(10).index
        df_g = dff[dff['genero_principal'].isin(top_g)].groupby(['genero_principal', 'tipo_contenido'])['popularity'].mean().reset_index()
        fig_bar = px.bar(
            df_g, x='popularity', y='genero_principal', color='tipo_contenido', barmode='group', orientation='h',
            title='Popularidad Promedio en el Top 10 de Géneros',
            labels={'popularity': 'Popularidad Promedio', 'genero_principal': 'Género Principal', 'tipo_contenido': 'Formato'},
            color_discrete_map=COLOR_MAP, template='plotly_white'
        )
        fig_bar.update_layout(yaxis={'categoryorder': 'total ascending'})

        df_sc = dff[(dff['vote_average'] > 0) & (dff['popularity'] < 400)]
        fig_sc = px.scatter(
            df_sc, x='vote_average', y='popularity', color='tipo_contenido', opacity=0.45,
            hover_data=['title', 'genero_principal', 'pais_principal'],
            title='Matriz de Calidad (Calificación) vs. Tracción (Popularidad)',
            labels={'vote_average': 'Calificación Promedio (1-10)', 'popularity': 'Índice de Popularidad', 'tipo_contenido': 'Formato'},
            color_discrete_map=COLOR_MAP, template='plotly_white'
        )
        contenido = html.Div(style={'display': 'grid', 'gridTemplateColumns': '1fr 1fr', 'gap': '18px'}, children=[
            html.Div(style={'backgroundColor': 'white', 'padding': '15px', 'borderRadius': '10px'}, children=[dcc.Graph(figure=fig_bar)]),
            html.Div(style={'backgroundColor': 'white', 'padding': '15px', 'borderRadius': '10px'}, children=[dcc.Graph(figure=fig_sc)])
        ])
    else:
        df_p = dff[dff['pais_principal'] != 'Unknown'].groupby(['pais_principal', 'tipo_contenido']).agg(
            cantidad=('show_id', 'count'),
            popularidad=('popularity', 'mean')
        ).reset_index()
        top_p = dff[dff['pais_principal'] != 'Unknown']['pais_principal'].value_counts().head(10).index
        df_p = df_p[df_p['pais_principal'].isin(top_p)]
        fig_pais = px.bar(
            df_p, x='cantidad', y='pais_principal', color='tipo_contenido', barmode='stack', orientation='h',
            title='Top 10 Países Productores por Volumen de Catálogo',
            labels={'cantidad': 'Cantidad de Títulos', 'pais_principal': 'País Productor', 'tipo_contenido': 'Formato'},
            color_discrete_map=COLOR_MAP, template='plotly_white'
        )
        fig_pais.update_layout(yaxis={'categoryorder': 'total ascending'})

        fig_pais_pop = px.bar(
            df_p, x='popularidad', y='pais_principal', color='tipo_contenido', barmode='group', orientation='h',
            title='Engagement (Popularidad Media) por País de Origen',
            labels={'popularidad': 'Popularidad Promedio', 'pais_principal': 'País Productor', 'tipo_contenido': 'Formato'},
            color_discrete_map=COLOR_MAP, template='plotly_white'
        )
        fig_pais_pop.update_layout(yaxis={'categoryorder': 'total ascending'})
        contenido = html.Div(style={'display': 'grid', 'gridTemplateColumns': '1fr 1fr', 'gap': '18px'}, children=[
            html.Div(style={'backgroundColor': 'white', 'padding': '15px', 'borderRadius': '10px'}, children=[dcc.Graph(figure=fig_pais)]),
            html.Div(style={'backgroundColor': 'white', 'padding': '15px', 'borderRadius': '10px'}, children=[dcc.Graph(figure=fig_pais_pop)])
        ])

    return kpis, contenido

if __name__ == '__main__':
    app.run(debug=True, port=8050)
```
